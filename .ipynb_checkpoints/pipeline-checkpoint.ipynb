```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline

# 1. Load a small subset of a real text dataset (Categories: Tech vs. Sports)
categories = ["comp.graphics", "rec.sport.baseball"]
print("Loading dataset...")
data = fetch_20newsgroups(
    subset="train", categories=categories, remove=("headers", "footers", "quotes")
)

X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

# 2. Build the Pipeline
pipeline = Pipeline(
    [
        ("vect", CountVectorizer(stop_words="english")),
        ("tfidf", TfidfTransformer()),
        ("clf", LogisticRegression()),
    ]
)

# 3. Define the Hyperparameter Grid to search over
param_grid = {
    "vect__max_df": [0.8, 1.0],  # Ignore very common words
    "vect__ngram_range": [(1, 1), (1, 2)],  # Single words vs word pairs
    "clf__C": [0.1, 1.0, 10.0],  # Regularization strength
}

# 4. Set up GridSearchCV (3-fold cross-validation)
grid_search = GridSearchCV(pipeline, param_grid, cv=3, n_jobs=-1, verbose=1)

# 5. Fit the entire workflow with one line!
print("\nRunning Grid Search...")
grid_search.fit(X_train, y_train)

# 6. Check the results
print("\n--- RESULTS ---")
print("Best Cross-Validation Score:", round(grid_search.best_score_, 4))
print("Best Parameters Found:")
for param, value in grid_search.best_params_.items():
    print(f"  {param}: {value}")

# 7. Test it on unseen data automatically using the best model found
test_score = grid_search.score(X_test, y_test)
print(f"Test Accuracy with Best Model: {round(test_score * 100, 2)}%")
```
